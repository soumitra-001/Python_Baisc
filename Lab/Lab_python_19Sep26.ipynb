```python
# Q1. Write a program to print the following using while loop First 10 Odd numbers First 10 Even numbers First 10 Whole numbers First 10 Natural numbers.
print("Odd Number:")
counter = 0
i = 1
while True:
  print(i)
  i += 2
  counter += 1
  if(counter==10):
    break

print("even Number:")
counter = 0
i = 2
while True:
  print(i)
  i += 2
  counter += 1
  if(counter==10):
    break
```
