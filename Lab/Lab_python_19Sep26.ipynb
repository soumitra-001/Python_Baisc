```python
# Q1. Write a program to print the following using while loop First 10 Odd numbers First 10 Even numbers First 10 Whole numbers First 10 Natural numbers.
print("Odd Number:")
counter = 0
i = 1
while True:
  print(i)
  i += 2
  counter += 1
  if(counter==10):
    break

print("even Number:")
counter = 0
i = 2
while True:
  print(i)
  i += 2
  counter += 1
  if(counter==10):
    break
```

```python
# Q6. Write a program to find the sum of the first 10 even numbers (2 + 4 + 6 + ... + 20) using while loop.
print("Sum of first 10 even Number:")
sum=0
counter = 0
i = 2
while True:
  sum = sum +i
  i += 2
  counter += 1
  if(counter==10):
    break
print(sum)
```

```python
# Q7. Write a program that takes a number as input from the user and prints its multiplication table from 1 to 10 using while loop.
num = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(num, "x", i, "=", num * i)
    i += 1
```
