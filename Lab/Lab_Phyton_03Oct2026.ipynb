```python
fruits = {"apple", "banana", "mango"}

print(fruits)
```

```python
# add element in set
fruits = {"apple", "banana", "mango"}

fruits.add("orange")

print(fruits)
```

```python
# add multiple elements in the set
fruits = {"apple", "banana", "mango"}

fruits.update(["orange", "grapes", "watermelon"])

print(fruits)
```

```python
# remove the elements

fruits.remove("banana")

print(fruits)
```

```python
# discard the elements

fruits = {"apple", "banana", "mango", "orange"}

fruits.discard("banana")

print(fruits)
```

```python
fruits = {"apple", "banana"}

fruits.discard("mango")

print(fruits)

fruits = {"apple", "banana"}

fruits.remove("mango")

print(fruits)
```
