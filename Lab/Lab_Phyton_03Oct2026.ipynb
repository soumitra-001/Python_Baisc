```python
'''
1.Scenario: Each hospital maintains a set of medicines in stock. You must:
a)Identify medicines common to all hospitals.
b) Determine medicines unique to each hospital.
c) Find if any hospital’s stock is a subset of another.
d) Create a report showing total unique medicines available across all hospitals.
hospital_A = {"Paracetamol", "Amoxicillin", "Ibuprofen"}
hospital_B = {"Ibuprofen", "Metformin", "Paracetamol"}
hospital_C = {"Metformin", "Paracetamol", "Azithromycin"}
'''

# a)Identify medicines common to all hospitals.
hospital_A = {"Paracetamol", "Amoxicillin", "Ibuprofen"}
hospital_B = {"Ibuprofen", "Metformin", "Paracetamol"}
hospital_C = {"Metformin", "Paracetamol", "Azithromycin"}

common = hospital_A & hospital_B & hospital_C

print("Common medicines:", common)
```

```python
#b) Determine medicines unique to each hospital.
unique_A = hospital_A - hospital_B - hospital_C
unique_B = hospital_B - hospital_A - hospital_C
unique_C = hospital_C - hospital_A - hospital_B

print("Unique to Hospital A:", unique_A)
print("Unique to Hospital B:", unique_B)
print("Unique to Hospital C:", unique_C)
```

```python
# c) Find if any hospital’s stock is a subset of another.
print("A subset of B:", hospital_A.issubset(hospital_B))
print("A subset of C:", hospital_A.issubset(hospital_C))

print("B subset of A:", hospital_B.issubset(hospital_A))
print("B subset of C:", hospital_B.issubset(hospital_C))

print("C subset of A:", hospital_C.issubset(hospital_A))
print("C subset of B:", hospital_C.issubset(hospital_B))
```

```python
#d) Create a report showing total unique medicines available across all hospitals.
all_medicines = hospital_A | hospital_B | hospital_C

print("All unique medicines:", all_medicines)
print("Total unique medicines:", len(all_medicines))
```

```python
'''
2.Scenario: Three departments share access to secure resources: Admin, Network, and Audit. You must:
a)Identify users with full (all three) access.
b)Detect users with conflicting privileges,i.e., those who have access to Admin but not Audit.
c) Determine users with no overlap in any department.
d) Print all unique users and the count.
'''
admin = {"Alice", "Bob", "Charlie", "David"}
network = {"Bob", "Charlie", "Eve", "Frank"}
audit = {"Alice", "Charlie", "Frank", "Grace"}
```

```python
# a)Identify users with full (all three) access.
full_access = admin & network & audit

print("Full access:", full_access)
```

```python
# b)Detect users with conflicting privileges,i.e., those who have access to Admin but not Audit.
conflicting  = admin - audit
print("Admin but not Audit:", conflicting)
```

```python
# c) Determine users with no overlap in any department.
only_admin = admin - network - audit
only_network = network - admin - audit
only_audit = audit - admin - network

no_overlap = only_admin | only_network | only_audit

print("No overlap:", no_overlap)
```

```python
# d) Print all unique users and the count.
all_users = admin | network | audit

print("All unique users:", all_users)
print("Total users:", len(all_users))
```

```python
'''
3.Scenario: Authors submit papers for Track A, Track B, and Track C. Some authors submit to multiple tracks, some only to one. You must:
a) Identify authors who submitted to all tracks.
b) Display authors who submitted to exactly two tracks.
c) Print authors who submitted only once.
d) Calculate and print the acceptance overlap percentage between Track A and Track B.
e) If Track A is a subset of Track B → print a special message.
Given:
track_A = {"P1", "P2", "P3", "P4"}
track_B = {"P3", "P4", "P5"}
track_C = {"P2", "P3", "P6"}
'''
track_A = {"P1", "P2", "P3", "P4"}
track_B = {"P3", "P4", "P5"}
track_C = {"P2", "P3", "P6"}
```

```python
# a) Identify authors who submitted to all tracks.
all_tracks = track_A & track_B & track_C

print("Submitted to all tracks:", all_tracks)
```

```python
#b) Display authors who submitted to exactly two tracks.
A_B = (track_A & track_B) - track_C
A_C = (track_A & track_C) - track_B
B_C = (track_B & track_C) - track_A

exactly_two = A_B | A_C | B_C

print("Submitted to exactly two tracks:", exactly_two)
```

```python
only_A = track_A - track_B - track_C
only_B = track_B - track_A - track_C
only_C = track_C - track_A - track_B

submitted_once = only_A | only_B | only_C

print("Submitted only once:", submitted_once)
```

```python
# d) Calculate and print the acceptance overlap percentage between Track A and Track B.
overlap_percentage = len(track_A & track_B) / len(track_A | track_B) * 100

print("Acceptance overlap percentage:", overlap_percentage, "%")
```

```python
# e) If Track A is a subset of Track B → print a special message.
if track_A.issubset(track_B):
    print("Track A is a subset of Track B")
```

```python
'''
4.Scenario: A sports federation has shortlisted players based on three games: Cricket, Football, Hockey. You must:
a) Find athletes selected for multiple sports.
b) Display who qualified for exactly one sport.
c) Compute the final squad (union of all sets).
d) Mark elite players — those who appear in all three lists.
Given:
cricket = {"A", "B", "C", "D"}
football = {"B", "C", "E"}
hockey = {"C", "D", "F"}
'''
cricket = {"A", "B", "C", "D"}
football = {"B", "C", "E"}
hockey = {"C", "D", "F"}
```

```python
# a) Find athletes selected for multiple sports.
multiple = (cricket & football) | (cricket & hockey) | (football & hockey)

print("Selected for multiple sports:", multiple)
```

```python
# b) Display who qualified for exactly one sport.
only_cricket = cricket - football - hockey
only_football = football - cricket - hockey
only_hockey = hockey - cricket - football

exactly_one = only_cricket | only_football | only_hockey

print("Qualified for exactly one sport:", exactly_one)
```

```python
# c) Compute the final squad (union of all sets).
final_squad = cricket | football | hockey

print("Final squad:", final_squad)
print("Total squad:", len(final_squad))
```

```python
# d) Mark elite players — those who appear in all three lists.
elite = cricket & football & hockey

print("Elite players:", elite)
```
