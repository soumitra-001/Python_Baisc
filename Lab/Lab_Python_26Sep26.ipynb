```python
# 1. Calculate the total sum of all integers in a list and find the arithmetic mean (average). Numbers: [10, 20, 30, 40, 50]
numbers = [10, 20, 30, 40, 50]

sum = 0
i = 0
while i < len(numbers):
    sum += numbers[i]
    i += 1

print("Sum of List is :",sum)
print("Mean of the List is :",sum/len(numbers))
```

```python
# 3. Multiply every number in a list together to find the total product. Factors: [2, 3, 5, 7]
factors = [2, 3, 5, 7]

prod = 1
i = 0
while i < len(factors):
    prod *= factors[i]
    i += 1

print("Sum of List is :",prod)
```

```python
# 4. Given a list of integers, iterate through the items and count how many are even and how many are odd. Numbers: [10, 21, 93, 11]
numbers= [10, 21, 93, 11]

count = 0
i = 0
while i < len(numbers):
    if(numbers[i] % 2 != 0):
      count += 1
    i += 1

print("Count of odd digit in list :",count)
```

```python
# 10. Write a check to see if a certain value exists within a list and print a message based on the result. Inventory: ["Laptop", "Mouse", "Monitor", "Keyboard"]; Target: "Tablet"

inventory= ["Laptop", "Mouse", "Monitor", "Keyboard"]
target= "Tablet"

i = 0
while i < len(inventory):
    if(inventory[i].lower() == target.lower()):
      print(target, "available in the inventory")
    i += 1

print(target,"not available in the inventory")
```

```python
data = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

result = []

i = 0

while i < len(data):
    j = 0

    while j < len(data[i]):
        result.append(data[i][j])
        j += 1

    i += 1

print(result)
```
