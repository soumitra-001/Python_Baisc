ATM System (BankAccount inheritance)

Build a minimal banking module supporting BankAccount (base), SavingsAccount, and CurrentAccount. Include deposit/withdrawal rules and behavior differences.

BankAccount has holder, balance; deposit(amount) and withdraw(amount) implemented with validation.

SavingsAccount adds apply_interest(rate) that increases balance by balance * rate.

CurrentAccount charges a withdrawal_fee per withdrawal.

Provide statement() to display details.




```python
class Vehicle:
  def __init__(self, base_fare):
    self.base_fare = base_fare
  def per_km_rate(self):
    pass
  def calculate_fare(self, km):
    if km <0:
      raise ValueError("Distance must be non-negative")

    fare = self.base_fare + self.per_km_rate() *km
    return fare

class Car(Vehicle):
  def per_km_rate(self):
    return 15

class Bike(Vehicle):
  def per_km_rate(self):
    return 8
class Auto(Vehicle):
  def per_km_rate(self):
    return 10

def get_ride_fare(vehicle, km):
  return vehicle.calculate_fare(km)


car = Car(base_fare=50)
bike = Bike(base_fare=20)
auto = Auto(base_fare=30)

print(get_ride_fare(car, 5))#50+15*5= 125
print(get_ride_fare(bike, 5))
print(get_ride_fare(auto, 5))
```

```python
class BankAccount:
  def __init__(self, holder, balance=0):
    self.holder = holder
    self.balance = balance
  def deposit(self, amount):
    if amount <= 0:
      raise ValueError("Amount must be positive")
    self.balance+=amount # self.balance= self.balance +amount
  def withdraw(self, amount):
    if amount <= 0:
      raise ValueError("Amount must be positive")
    if amount > self.balance: # 515>2000
      raise ValueError("Insufficient funds")
    self.balance -= amount
# 2000-515= 1485
  def statment(self):
    print(self.holder, self.balance)

class SavingAccount(BankAccount):
  def apply_interest(self, rate):
    self.balance += self.balance * rate

class CurrentAccount(BankAccount):
  def __init__(self, holder, balance=0, withdrawal_fee=100):
    super().__init__(holder, balance)
    self.withdrawal_fee = withdrawal_fee
  def withdraw(self, amount):
    super().withdraw(amount + self.withdrawal_fee)
```

```python
s = SavingAccount("amit", 1000)
s.apply_interest(0.05)
s.statment()
# 1000+1000 *0.05
```

```python
c = CurrentAccount("ravi", 2000, withdrawal_fee=15)
c.withdraw(500)
c.statment()
```
