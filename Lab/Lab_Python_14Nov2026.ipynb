```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Soumitra", 22)

print(s1.name)
print(s1.age)
```

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student Name: {self.name}"

s1 = Student("Rahul")

print(s1)
```

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student('{self.name}')"

s1 = Student("Rahul")

print(repr(s1))
```

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

t1 = Team(["Rahul", "Amit", "Priya"])

print(len(t1))
```

```python
class Numbers:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

n1 = Numbers(10)
n2 = Numbers(20)

print(n1 + n2)
```

```python
class Numbers:
    def __init__(self, value):
        self.value = value

    def __sub__(self, other):
        return self.value - other.value

n1 = Numbers(50)
n2 = Numbers(20)

print(n1 - n2)
```

```python
class Numbers:
    def __init__(self, value):
        self.value = value

    def __mul__(self, other):
        return self.value * other.value

n1 = Numbers(5)
n2 = Numbers(4)

print(n1 * n2)
```

```python
class Student:
    def __init__(self, age):
        self.age = age

    def __eq__(self, other):
        return self.age == other.age

s1 = Student(22)
s2 = Student(22)

print(s1 == s2)
```

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __lt__(self, other):
        return self.marks < other.marks

s1 = Student(70)
s2 = Student(90)

print(s1 < s2)
```

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __gt__(self, other):
        return self.marks > other.marks

s1 = Student(90)
s2 = Student(70)

print(s1 > s2)
```

```python
class Greeting:
    def __call__(self, name):
        print("Hello", name)

obj = Greeting()

obj("Rahul")
```
