```python
'''
5. Student Academic Analyzer: Store a student’s name (string), roll number (int), and marks of three subjects (float). Print CGPA rounded to 2 decimals and display the data type of each variable.
Calculate CGPA assuming: CGPA = (average marks) / 10
'''
name = "Rahul"
roll_number = 101
marks1 = 85.0
marks2 = 90.0
marks3 = 80.0

# Calculate average marks
average = (marks1 + marks2 + marks3) / 3

# Calculate CGPA
cgpa = average / 10

# Display details
print("Student Details")
print("Name:", name)
print("Roll Number:", roll_number)
print(f"CGPA: {cgpa:.2f}")

# Display data types
print("\nData Types")
print("Name:", type(name))
print("Roll Number:", type(roll_number))
print("Marks 1:", type(marks1))
print("Marks 2:", type(marks2))
print("Marks 3:", type(marks3))
```
