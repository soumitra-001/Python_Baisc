1. os Module
- Used to interact with the operating system.
- Helps create, delete, and manage directories.
- Provides functions to work with files and system information.

Common functions:
- os.getcwd() – Get the current working directory.
- os.mkdir() – Create a directory.
- os.listdir() – List files and folders.
- os.remove() – Delete a file.
- os.rmdir() – Delete an empty directory.

2. os.path Module
- Used to work with file and directory paths.
- Helps check whether a file or directory exists.
- Provides methods to join and extract path information.

Common functions:
- os.path.exists() – Check whether a path exists.
- os.path.isfile() – Check whether a path is a file.
- os.path.isdir() – Check whether a path is a directory.
- os.path.join() – Join two or more paths.
- os.path.basename() – Get the file name from a path.
- os.path.dirname() – Get the directory name from a path.

```python
# Normalise a messy path
print(os.path.normpath("data//raw/../clear/file.txt"))
```

```python
path = "/home/user/documents/report.txt"

print("File name:", os.path.basename(path))
print("Directory name:", os.path.dirname(path))
print("File name and extension:", os.path.splitext(path))
```

3. pathlib Module
- Provides an object-oriented way to work with file and directory paths.
- Makes path handling easier and more readable.
- Can create, read, write, and delete files.

Common functions:
- Path.cwd() – Get the current working directory.
- Path.mkdir() – Create a directory.
- Path.exists() – Check whether a path exists.
- Path.is_file() – Check whether it is a file.
- Path.is_dir() – Check whether it is a directory.
- Path.unlink() – Delete a file.
- Path.rmdir() – Delete an empty directory.
