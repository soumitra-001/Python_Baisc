File Handling and file system operation in python

os

1. os Module
- Used to interact with the operating system.
- Helps create, delete, and manage directories.
- Provides functions to work with files and system information.

Common functions:
- os.getcwd() – Get the current working directory.
- os.mkdir() – Create a directory.
- os.listdir() – List files and folders.
- os.remove() – Delete a file.
- os.rmdir() – Delete an empty directory.

```python
# check Operating system name
import os
print(os.name)
```

```python
#get the file details
with open("info.txt","w") as f:
  f.write("hello")

info=os.stat("info.txt")
print(info.st_size)#size in bytes
print(info.st_mtime) #last time of modifiedtime
```

```python
#showing last modified time in readable format

import time
print(time.ctime(os.path.getmtime("info.txt")))
```
