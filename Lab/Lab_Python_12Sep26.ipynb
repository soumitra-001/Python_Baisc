```python
# 6. Ask the user to enter a password. If its length is greater than 8, Password must contain at least one number (0-9), Password must contain at least one uppercase letter (A-Z), Password must contain at least one lowercase letter (a-z), Password must contain at least one special character (!@#$%^&*) then print "Strong Password".

password = input("Enter the password: ")

if (len(password) > 8
    and any(ch.isdigit() for ch in password)
    and any(ch.isupper() for ch in password)
    and any(ch.islower() for ch in password)
    and any(ch in "!@#$%^&*" for ch in password)):

    print("Strong Password")
else:
    print("Weak Password")
```

```python
# Q9: Write a Python program to create a simple menu-driven calculator that accepts two numbers and an arithmetic operator (+, -, *, /, %) from the user, then prints the calculated answer.
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /, %): ")

if operator == "+":
    print("Answer:", num1 + num2)

elif operator == "-":
    print("Answer:", num1 - num2)

elif operator == "*":
    print("Answer:", num1 * num2)

elif operator == "/":
    if num2 != 0:
        print("Answer:", num1 / num2)
    else:
        print("Cannot divide by zero")

elif operator == "%":
    if num2 != 0:
        print("Answer:", num1 % num2)
    else:
        print("Cannot divide by zero")

else:
    print("Invalid operator")
```

```python
'''
Q10: Write a Python program to accept employee details including Age, Sex ('M' or 'F'), and Number of Days worked, then compute the total wage using the following csv:
Age Group,Sex,Wage / Day (₹)
---------------------------------
">=18 and <30",Male ('M'),700
">=18 and <30",Female ('F'),750
">=30 and <=40",Male ('M'),800
">=30 and <=40",Female ('F'),850
'''
age = int(input("Enter employee age: "))
sex = input("Enter sex (M/F): ")
days = int(input("Enter number of days worked: "))

if age >= 18 and age < 30:
    if sex == "M":
        wage = 700
    elif sex == "F":
        wage = 750
    else:
        wage = 0
        print("Invalid sex")

elif age >= 30 and age <= 40:
    if sex == "M":
        wage = 800
    elif sex == "F":
        wage = 850
    else:
        wage = 0
        print("Invalid sex")

else:
    wage = 0
    print("Age is not within the given range")

if wage > 0:
    total_wage = wage * days
    print("Total Wage: ₹", total_wage)
```
