```python
# 6. Ask the user to enter a password. If its length is greater than 8, Password must contain at least one number (0-9), Password must contain at least one uppercase letter (A-Z), Password must contain at least one lowercase letter (a-z), Password must contain at least one special character (!@#$%^&*) then print "Strong Password".

password = input("Enter the password: ")

if (len(password) > 8
    and any(ch.isdigit() for ch in password)
    and any(ch.isupper() for ch in password)
    and any(ch.islower() for ch in password)
    and any(ch in "!@#$%^&*" for ch in password)):

    print("Strong Password")
else:
    print("Weak Password")
```
