```python
# 4. Days of the week. A school timetable app must never change the names of the days. Store the 7 days in a tuple. Ask the user for a day number (1 to 7) and
# print the day name. Print an error if the number is invalid.
days = ("Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday")

num = int(input("Enter a day number (1-7): "))

if num >= 1 and num <= 7:
    print("Day:", days[num - 1])
else:
    print("Error: Invalid day number!")
```

```python
# 5. Map coordinates. A delivery app stores each location as a (latitude, longitude) pair. Given home = (25.56, 85.04) and office = (25.61, 85.14), unpack both and print
#how far apart they are in latitude and in longitude (absolute difference).

home = (25.56, 85.04)
office = (25.61, 85.14)

# Unpack the tuples
home_lat, home_lon = home
office_lat, office_lon = office

# Calculate absolute differences
lat_diff = abs(office_lat - home_lat)
lon_diff = abs(office_lon - home_lon)

print("Latitude difference:", lat_diff)
print("Longitude difference:", lon_diff)
```

```python
'''
Problem 1: GPS Location System
Scenario: A travel mapping app uses tuples to store GPS coordinates of different landmarks to ensure data integrity — once stored, the coordinates cannot be modified directly (immutability keeps records authentic). # Locations: (latitude, longitude, landmark_name, city) locations = [ (28.6139, 77.2090, "India Gate", "New Delhi"), (19.0760, 72.8777, "Gateway of India", "Mumbai"), (13.0827, 80.2707, "Marina Beach", "Chennai"), (34.0522, -118.2437, "Hollywood Sign", "Los Angeles"), (40.7128, -74.0060, "Statue of Liberty", "New York") ] Task A: Location Dictionary
Create a dictionary where tuple coordinates are keys and landmark names are values.
Task B: Update Coordinates
It’s discovered that the stored coordinates for Marina Beach are slightly off. The correct coordinates are (13.0500, 80.2824). Create a new locations list with this correction — don’t modify the existing tuples directly, since tuples are immutable.
Task C: Find Nearby Locations
FindNearby location based on Latitude.
Southern India: 12.0°N to 20.0°N Northern India: 25.0°N to 35.0°N USA: 30.0°N to 45.0°N' Example: find_locations_in_latitude_range(locations, 12.0, 20.0) → should return all landmarks located in southern India.Task D: Add New Location
Add “Taj Mahal” in Agra with coordinates (27.1751, 78.0421) to the locations list — again, do this without modifying existing tuples, maintaining immutability.
'''

# GPS Location System

locations = [
    (28.6139, 77.2090, "India Gate", "New Delhi"),
    (19.0760, 72.8777, "Gateway of India", "Mumbai"),
    (13.0827, 80.2707, "Marina Beach", "Chennai"),
    (34.0522, -118.2437, "Hollywood Sign", "Los Angeles"),
    (40.7128, -74.0060, "Statue of Liberty", "New York")
]


# Task A: Location Dictionary
location_dict = {}

for lat, lon, landmark, city in locations:
    location_dict[(lat, lon)] = landmark

print("Task A: Location Dictionary")
print(location_dict)


# Task B: Update Coordinates
updated_locations = []

for lat, lon, landmark, city in locations:
    if landmark == "Marina Beach":
        updated_locations.append(
            (13.0500, 80.2824, landmark, city)
        )
    else:
        updated_locations.append((lat, lon, landmark, city))

print("\nTask B: Updated Locations")
for location in updated_locations:
    print(location)


# Task C: Find Nearby Locations Based on Latitude
def find_locations_in_latitude_range(locations, min_lat, max_lat):
    result = []

    for lat, lon, landmark, city in locations:
        if min_lat <= lat <= max_lat:
            result.append((landmark, city))

    return result


print("\nSouthern India:")
print(find_locations_in_latitude_range(
    updated_locations, 12.0, 20.0
))

print("\nNorthern India:")
print(find_locations_in_latitude_range(
    updated_locations, 25.0, 35.0
))

print("\nUSA:")
print(find_locations_in_latitude_range(
    updated_locations, 30.0, 45.0
))


# Task D: Add New Location
new_location = (27.1751, 78.0421, "Taj Mahal", "Agra")

final_locations = updated_locations + [new_location]

print("\nTask D: Final Locations")
for location in final_locations:
    print(location)
```
