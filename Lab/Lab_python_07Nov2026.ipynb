**Develop a Python program to manage cricket player information.**

Create a class named CricketPlayer with the following attributes:
- Player name (public).
- Team name (public).
- Jersey number (protected).
- Total runs scored (private).
- Total matches played (private).
Implement the following methods:
1. A constructor to initialize all player details.
2. A method to display player information.
3. A getter method to retrieve total runs.
4. A setter method to update total runs, ensuring that the value cannot be negative.
5. A method to calculate the player's batting average.
Condition: The private attributes should not be directly accessible or modified from outside the class.

```python

class CricketPlayer:

    def __init__(self, player_name, team_name, jersey_number, total_runs, total_matches):
        # Public attributes
        self.player_name = player_name
        self.team_name = team_name

        # Protected attribute
        self._jersey_number = jersey_number

        # Private attributes
        self.__total_runs = total_runs
        self.__total_matches = total_matches

    # Display player information
    def display_info(self):
        print("Player Name:", self.player_name)
        print("Team Name:", self.team_name)
        print("Jersey Number:", self._jersey_number)
        print("Total Runs:", self.__total_runs)
        print("Total Matches:", self.__total_matches)

    # Getter method
    def get_total_runs(self):
        return self.__total_runs

    # Setter method
    def set_total_runs(self, runs):
        if runs >= 0:
            self.__total_runs = runs
            print("Total runs updated successfully.")
        else:
            print("Error: Runs cannot be negative.")

    # Calculate batting average
    def batting_average(self):
        if self.__total_matches > 0:
            return self.__total_runs / self.__total_matches
        else:
            return 0


# Creating an object
player1 = CricketPlayer("Virat Kohli", "India", 18, 14000, 300)

# Display player details
player1.display_info()

# Retrieve total runs using getter
print("\nTotal Runs:", player1.get_total_runs())

# Update total runs using setter
player1.set_total_runs(15000)

# Calculate batting average
print("Batting Average:", player1.batting_average())
```

**Develop an Automobile Management System using inheritance.**

Create a parent class named Automobile with the following attributes:
* Brand
* Model
* Manufacturing year
* Fuel type
* Implement a constructor and a method named display_info().

Create three child classes:
* Car
* Bike
* Truck
Each child class should inherit the common properties and methods from the parent class.
Additional requirements:
Car should have a number_of_doors attribute.
Bike should have an engine_cc attribute.
Truck should have a load_capacity attribute.
Use super() to initialize the parent class attributes.
Each child class should have its own method to display its specific information.



```python

# Parent class
class Automobile:

    def __init__(self, brand, model, manufacturing_year, fuel_type):
        self.brand = brand
        self.model = model
        self.manufacturing_year = manufacturing_year
        self.fuel_type = fuel_type

    # Common method
    def display_info(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Manufacturing Year:", self.manufacturing_year)
        print("Fuel Type:", self.fuel_type)


# Child class 1
class Car(Automobile):

    def __init__(self, brand, model, manufacturing_year, fuel_type, number_of_doors):

        super().__init__(brand, model, manufacturing_year, fuel_type)

        self.number_of_doors = number_of_doors

    def display_car_info(self):
        print("\n----- CAR INFORMATION -----")
        self.display_info()
        print("Number of Doors:", self.number_of_doors)


# Child class 2
class Bike(Automobile):

    def __init__(self, brand, model, manufacturing_year, fuel_type, engine_cc):

        super().__init__(brand, model, manufacturing_year, fuel_type)

        self.engine_cc = engine_cc

    def display_bike_info(self):
        print("\n----- BIKE INFORMATION -----")
        self.display_info()
        print("Engine CC:", self.engine_cc)


# Child class 3
class Truck(Automobile):

    def __init__(self, brand, model, manufacturing_year, fuel_type, load_capacity):

        super().__init__(brand, model, manufacturing_year, fuel_type)

        self.load_capacity = load_capacity

    def display_truck_info(self):
        print("\n----- TRUCK INFORMATION -----")
        self.display_info()
        print("Load Capacity:", self.load_capacity)


# Creating objects
car1 = Car("Toyota", "Fortuner", 2024, "Diesel", 4)

bike1 = Bike("Royal Enfield", "Classic 350", 2023, "Petrol", 350)

truck1 = Truck("Tata", "Prima", 2022, "Diesel", "20 Tons")


# Displaying information
car1.display_car_info()

bike1.display_bike_info()

truck1.display_truck_info()
```


A marketing company wants to develop a system to manage different types of marketing campaigns.
The company uses three marketing strategies:
* Email Marketing.
* Social Media Marketing.
* SMS Marketing.
Create an abstract class named MarketingCampaign with:
* Campaign name.
* Budget.
Define the following abstract methods:
* run_campaign()
* calculate_reach()
Create three child classes:
* EmailMarketing
* SocialMediaMarketing
* SMSMarketing
Each child class must implement both abstract methods.
Assume the following reach calculations:
Marketing type	Reach calculation
Email	Budget × 100
Social Media	Budget × 200
SMS	Budget × 150
Finally, create objects of all three marketing strategies and execute their campaigns.


```python

from abc import ABC, abstractmethod


# Abstract Parent Class
class MarketingCampaign(ABC):

    def __init__(self, campaign_name, budget):
        self.campaign_name = campaign_name
        self.budget = budget

    @abstractmethod
    def run_campaign(self):
        pass

    @abstractmethod
    def calculate_reach(self):
        pass


# Child Class 1
class EmailMarketing(MarketingCampaign):

    def run_campaign(self):
        print("Email Marketing campaign is running.")

    def calculate_reach(self):
        return self.budget * 100


# Child Class 2
class SocialMediaMarketing(MarketingCampaign):

    def run_campaign(self):
        print("Social Media Marketing campaign is running.")

    def calculate_reach(self):
        return self.budget * 200


# Child Class 3
class SMSMarketing(MarketingCampaign):

    def run_campaign(self):
        print("SMS Marketing campaign is running.")

    def calculate_reach(self):
        return self.budget * 150


# Creating objects
email = EmailMarketing("Email Promotion", 1000)

social = SocialMediaMarketing("Instagram Promotion", 1000)

sms = SMSMarketing("SMS Promotion", 1000)


# Storing objects in a single list
campaigns = [email, social, sms]


# Executing campaigns
for campaign in campaigns:

    print("\nCampaign Name:", campaign.campaign_name)
    print("Budget:", campaign.budget)

    campaign.run_campaign()

    reach = campaign.calculate_reach()

    print("Expected Reach:", reach)

    print("----------------------------")
```

**Develop a Banking Management System to manage customer bank accounts.**
Create a class named BankAccount with the following attributes:
1. Account holder name (public).
2. Account number (protected).
3. Account balance (private).
4. Implement the following methods:
5. A constructor to initialize account details.
6. A getter method to retrieve the account balance.
7. A deposit() method to deposit money.
8. A withdraw() method to withdraw money.
9. A display_account() method to display account information.
Conditions:
* The account balance must not be directly accessible or modified from outside the class.
* Deposited amounts must be positive.
* Withdrawals must not exceed the available balance.
* The account number should follow the protected naming convention.


```python

class BankAccount:

    # Constructor
    def __init__(self, holder_name, account_number, balance):

        # Public attribute
        self.holder_name = holder_name

        # Protected attribute
        self._account_number = account_number

        # Private attribute
        self.__balance = balance

    # Getter method
    def get_balance(self):
        return self.__balance

    # Deposit method
    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount
            print("Amount deposited successfully:", amount)
        else:
            print("Error: Deposit amount must be positive.")

    # Withdraw method
    def withdraw(self, amount):

        if amount <= 0:
            print("Error: Withdrawal amount must be positive.")

        elif amount <= self.__balance:
            self.__balance -= amount
            print("Amount withdrawn successfully:", amount)

        else:
            print("Error: Insufficient account balance.")

    # Display account information
    def display_account(self):

        print("\n----- ACCOUNT INFORMATION -----")
        print("Account Holder:", self.holder_name)
        print("Account Number:", self._account_number)
        print("Account Balance:", self.__balance)


# Creating an object
account1 = BankAccount("Soumitra", "1234567890", 10000)


# Display account information
account1.display_account()

# Retrieve balance using getter
print("\nCurrent Balance:", account1.get_balance())

# Deposit money
account1.deposit(5000)

# Withdraw money
account1.withdraw(2000)

# Display updated balance
print("\nUpdated Balance:", account1.get_balance())
```

**Develop a Hospital Management System using inheritance.**

1. A constructor to initialize these attributes.

Create three child classes:
1. Doctor

* Doctor should have a specialization and consultation fee.
* Nurse should have a department and shift.
* Patient should have a disease and room number.
* Use super() to initialize common attributes.
* Each child class should have its own method to display its additional information.

```python

class HospitalPerson:
    # Parent class
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    # Common method
    def display_details(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Gender:", self.gender)


class Doctor(HospitalPerson):
    # Child class
    def __init__(self, name, age, gender, specialization, consultation_fee):
        super().__init__(name, age, gender)
        self.specialization = specialization
        self.consultation_fee = consultation_fee

    def display_doctor_info(self):
        self.display_details()
        print("Specialization:", self.specialization)
        print("Consultation Fee:", self.consultation_fee)


class Nurse(HospitalPerson):
    # Child class
    def __init__(self, name, age, gender, department, shift):
        super().__init__(name, age, gender)
        self.department = department
        self.shift = shift

    def display_nurse_info(self):
        self.display_details()
        print("Department:", self.department)
        print("Shift:", self.shift)


class Patient(HospitalPerson):
    # Child class
    def __init__(self, name, age, gender, disease, room_number):
        super().__init__(name, age, gender)
        self.disease = disease
        self.room_number = room_number

    def display_patient_info(self):
        self.display_details()
        print("Disease:", self.disease)
        print("Room Number:", self.room_number)


# Creating Doctor object
doctor1 = Doctor("Dr. Rahul", 40, "Male", "Cardiology", 800)

# Creating Nurse object
nurse1 = Nurse("Priya", 28, "Female", "Emergency", "Night")

# Creating Patient object
patient1 = Patient("Amit", 35, "Male", "Diabetes", 205)


# Displaying information
print("----- DOCTOR DETAILS -----")
doctor1.display_doctor_info()

print("\n----- NURSE DETAILS -----")
nurse1.display_nurse_info()

print("\n----- PATIENT DETAILS -----")
patient1.display_patient_info()
```

**Develop an E-commerce Payment System that supports different payment methods.**

Create a parent class named Payment with a method called process_payment().

Create three child classes:
1. CreditCardPayment
2. UPIPayment
3. alletPayment
Each child class should override the process_payment() method.
Requirements:
* Credit card payment should display the card payment confirmation.

# Food Delivery Management System

Develop a **Food Delivery Management System** that supports different delivery services.

## Delivery Methods

The system supports three delivery methods:

1. Bike Delivery
2. Car Delivery
3. Bicycle Delivery

## Abstract Class: `Delivery`

Create an abstract class named `Delivery` with the following attributes:

- Order ID
- Customer name
- Distance (in kilometres)

Define two abstract methods:

- `calculate_delivery_charge()`
- `deliver_order()`

## Child Classes

Create three child classes:

1. `BikeDelivery`
2. `CarDelivery`
3. `BicycleDelivery`

Each child class must implement both abstract methods.

## Delivery Charge Calculations

Assume the following delivery charge calculations:

| Delivery Method | Charge Calculation |
|---|---|
| Bike | Distance × ₹10 |
| Car | Distance × ₹20 |
| Bicycle | Distance × ₹5 |

## Final Requirements

- Create objects of all three delivery classes.
- Store them in a list.
- Display the delivery charge.
- Execute the delivery operation.

## Objective

Demonstrate how **abstraction** defines a common interface while hiding the specific delivery implementation.

```python
from abc import ABC, abstractmethod


# Abstract Parent Class
class Delivery(ABC):

    def __init__(self, order_id, customer_name, distance):
        self.order_id = order_id
        self.customer_name = customer_name
        self.distance = distance

    @abstractmethod
    def calculate_delivery_charge(self):
        pass

    @abstractmethod
    def deliver_order(self):
        pass


# Child Class 1: Bike Delivery
class BikeDelivery(Delivery):

    def calculate_delivery_charge(self):
        return self.distance * 10

    def deliver_order(self):
        print("Order delivered successfully using Bike.")


# Child Class 2: Car Delivery
class CarDelivery(Delivery):

    def calculate_delivery_charge(self):
        return self.distance * 20

    def deliver_order(self):
        print("Order delivered successfully using Car.")


# Child Class 3: Bicycle Delivery
class BicycleDelivery(Delivery):

    def calculate_delivery_charge(self):
        return self.distance * 5

    def deliver_order(self):
        print("Order delivered successfully using Bicycle.")


# Creating objects
delivery1 = BikeDelivery(101, "Rahul", 5)
delivery2 = CarDelivery(102, "Priya", 10)
delivery3 = BicycleDelivery(103, "Amit", 3)


# Storing objects in a list
deliveries = [delivery1, delivery2, delivery3]


# Processing all deliveries
for delivery in deliveries:

    print("\n------------------------------")
    print("Order ID:", delivery.order_id)
    print("Customer Name:", delivery.customer_name)
    print("Distance:", delivery.distance, "KM")

    print("Delivery Charge: ₹", delivery.calculate_delivery_charge())

    delivery.deliver_order()

    print("------------------------------")
```
