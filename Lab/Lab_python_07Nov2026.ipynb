```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)


emp1 = Employee("Soumitra", 30000)

emp1.display()
```

```python
class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Amount deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance")

    def display_balance(self):
        print("Account Holder:", self.name)
        print("Current Balance:", self.balance)


account1 = BankAccount("Soumitra", 10000)

account1.deposit(5000)
account1.withdraw(2000)
account1.display_balance()
```

```python
class Employee:

    company = "SmartRecon"  # Class attribute

    def __init__(self, name, salary):
        self.name = name       # Instance attribute
        self.salary = salary   # Instance attribute


emp1 = Employee("Soumitra", 30000)
emp2 = Employee("Rahul", 40000)

print(emp1.company)
print(emp2.company)

print(emp1.salary)
print(emp2.salary)
```

```python
class Employee:

    company = "SmartRecon"

    def __init__(self, name):
        self.name = name

    # Instance method
    def display(self):
        print(self.name)

    # Class method
    @classmethod
    def company_name(cls):
        print(cls.company)

    # Static method
    @staticmethod
    def message():
        print("Welcome to SmartRecon")


emp1 = Employee("Soumitra")

emp1.display()
Employee.company_name()
Employee.message()
```

```python
class Employee:

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Employee Name:", self.name)


class Developer(Employee):

    def coding(self):
        print(self.name, "is writing Python code")


dev1 = Developer("Soumitra")

dev1.display()
dev1.coding()
```
