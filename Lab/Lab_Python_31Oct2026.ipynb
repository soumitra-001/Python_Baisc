```python
'''
1.Imagine we’ve got a list of students with their names and scores, and we want to calculate something useful, like the average score of the top-performing students.
students = [
    {"name": "Anni", "score": 88},
    {"name": "Bhim", "score": 92},
    {"name": "Coco", "score": 70},
    {"name": "Dona", "score": 64},
    {"name": "Arjun", "score": 99}
]
Our goal: Find the average score of students who scored 75 or above.
'''

students = [
    {"name": "Anni", "score": 88},
    {"name": "Bhim", "score": 92},
    {"name": "Coco", "score": 70},
    {"name": "Dona", "score": 64},
    {"name": "Arjun", "score": 99}
]

top_students = list(filter(lambda x: x["score"] >= 75, students))

total = sum(map(lambda x: x["score"], top_students))

average = total / len(top_students)

print("Average Score:", average)
```

```python
# 2. Find the student with the highest score using lanbda function

highest = max(students, key=lambda x: x["score"])

print(highest)
```

```python
# 3.Sort the list of students by their scores in ascending order using a lambda function
sorted_students = sorted(students, key=lambda x: x["score"])

for student in sorted_students:
    print(student)
```

```python
# 4.Sort the list of students by their names alphabetically using a lambda function.
sorted_students = sorted(students, key=lambda x: x["name"])

for student in sorted_students:
    print(student)
```

```python
employees = [
    {"name": "Alice", "department": "Engineering", "salary": 80000, "years": 3},
    {"name": "Bob", "department": "Marketing", "salary": 65000, "years": 5},
    {"name": "Charlie", "department": "Engineering", "salary": 95000, "years": 8},
    {"name": "Diana", "department": "HR", "salary": 55000, "years": 2},
    {"name": "Eve", "department": "Engineering", "salary": 110000, "years": 10}
]
```

```python
# 1.Extract all employee names.
names = list(map(lambda x: x["name"], employees))

print(names)
```

```python
# 2.Filter employees earning more than 70,000.
high_salary = list(filter(lambda x: x["salary"] > 70000, employees))

print(high_salary)
```

```python
# 3.Get all employees from the Engineering department.
engineering = list(filter(lambda x: x["department"] == "Engineering", employees))

print(engineering)
```

```python
# 4.Calculate the total payroll (sum of all salaries).
total_payroll = sum(map(lambda x: x["salary"], employees))

print(total_payroll)
```

```python
# 5.Find the employee with the highest salary.
highest_salary = max(employees, key=lambda x: x["salary"])

print(highest_salary)
```

```python
# 6.Compute the average salary of all employees.
average_salary = sum(map(lambda x: x["salary"], employees)) / len(employees)

print(average_salary)
```

```python
# 7.Generate formatted strings like “Name (Dept): $Salary”.
formatted = list(map(
    lambda x: f'{x["name"]} ({x["department"]}): ${x["salary"]}',
    employees
))

print(formatted)
```

```python
# 8.Calculate total years of experience for Engineering employees.
engineering_years = sum(
    map(
        lambda x: x["years"],
        filter(lambda x: x["department"] == "Engineering", employees)
    )
)

print(engineering_years)
```

```python
# 9.Increase salary by 10% for employees with more than 5 years of experience.
updated_employees = list(map(
    lambda x: {
        **x,
        "salary": x["salary"] * 1.10
    } if x["years"] > 5 else x,
    employees
))

print(updated_employees)
```
