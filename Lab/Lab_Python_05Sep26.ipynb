```python
a = 10  # Binary: 1010  --> ... 8 4 2 1 --> 8 + 2 = 10
b = 4   # Binary: 0100  --> ... 8 4 2 1 --> 4

# 1. Bitwise AND (&)
# Sets bit to 1 ONLY if both bits are 1
#   1010  (10)
# & 0100  (4)
#   ----
#   0000  (0)
print(a & b)  # Output: 0


# 2. Bitwise OR (|)
# Sets bit to 1 if at least one bit is 1
#   1010  (10)
# | 0100  (4)
#   ----
#   1110  (8 + 4 + 2 + 0 = 14)
print(a | b)  # Output: 14


# 3. Bitwise NOT (~)
# Inverts all bits. Formula is always -(x + 1)
# ~10 becomes -(10 + 1)
print(~a)     # Output: -11


# 4. Bitwise XOR (^)
# Sets bit to 1 if the bits are DIFFERENT (one is 1, one is 0)
#   1010  (10)
# ^ 0100  (4)
#   ----
#   1110  (14)
print(a ^ b)  # Output: 14


# 5. Bitwise Right Shift (>>)
# Shifts bits to the right, dropping the last digits.
# Math Shortcut: Floor divide by 2^n -> 10 // 2^2 = 10 // 4 = 2
#   1010 (10)   00 --> 1010 --> 0010
#   >> 2
#   ----
#   0010  (2)
print(a >> 2) # Output: 2


# 6. Bitwise Left Shift (<<)
# Shifts bits to the left, adding zeros to the end.
# Math Shortcut: Multiply by 2^n -> 10 * 2^2 = 10 * 4 = 40
#   1010
#   << 2
#   ----
# 101000  (32 + 8 = 40)    --> 32 16 8 4 2 1 --> (32 + 8 = 40)
print(a << 2) # Output: 40
```
