### Question 1: Autonomous Vehicle Safety Controller

An autonomous vehicle stores sensor states using bit flags:

* `1` = Front obstacle detected
* `2` = Rear obstacle detected
* `4` = Lane departure detected
* `8` = Emergency brake active
* `16` = Manual override active

Given:

```python
status = 13
speed = 72
```

The vehicle must enter emergency mode if:

* Emergency brake is active, OR
* Front obstacle and lane departure are both detected while speed is greater than `60`, OR
* Rear obstacle is detected while manual override is not active.

Write a single logical expression using comparison, logical, and bitwise operators to determine whether emergency mode should activate.

````python
status = 13
speed = 72

result = ((status & 8) != 0) or (((status & 1) != 0) and ((status & 4) != 0) and (speed > 60)) or (((status & 2) != 0) and ((status & 16) == 0))
print(result)
````

### Question 2: Secure Data Center Authorization

A data center stores user privileges as:

* `1` = Basic access
* `2` = Server access
* `4` = Database access
* `8` = Administrator access
* `16` = Security clearance

Given:

```python
permissions = 22
security_level = 4
```

Access to the core server is granted if:

* Server access and security clearance are both enabled,
* AND security level is at least `3`,

OR

* Administrator access is enabled and security level is exactly `5`.

Write the complete Boolean expression that determines access.


````python
permissions = 22
security_level = 4

result = ((permissions & 2 != 0) and (permissions & 16 != 0) and security_level >= 3) or ((permissions & 8 != 0) and security_level == 5)

print(result)
````

Write a single expression that evaluates whether an emergency exists.
