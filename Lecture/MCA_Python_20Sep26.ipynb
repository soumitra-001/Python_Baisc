```python
for i in range(5):  #0, 1, 2, 3, 4
    print(i)
```

```python
for i in range(5):
  for j in range(5):
    print(i,j,end='  ')
```

```python
for i in range(5):
  for j in range(5):
    if i==j:
      continue
    print(i,j,end='  ')
```

```python
for i in range(10):  # condition is depended on the count
  i *= 2
  print(i,end=' ')
```

```python
for i in range(1, 6):  #range(start, stop)
    print(i)
```

```python
for i in range(1, 11, 2):    #range(start, stop, step)
    print(i)
```

```python
for i in range(10, 0, -1):
    print(i)
```

```python
# Even Number
for i in range(2, 21, 2):
    print(i, end=" ")
```
