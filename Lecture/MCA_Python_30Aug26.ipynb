```python
# Example 1: Counting characters in a string
text = "Hello World"
print(f"Length of '{text}': {len(text)}")
# Output: Length of 'Hello World': 11
```

```python
# Example 2: Counting specific characters
text = "banana"
print(f"Count of 'a' in '{text}': {text.count('a')}")
# Output: Count of 'a' in 'banana': 3
```

```python
sentence = "Python is awesome and Python is fun"
words = sentence.split()
print(f"Number of words: {len(words)}")
# Output: Number of words: 7
```

```python
# Example 1: upper() - converts all characters to uppercase
text = "hello world"
print(f"Original: {text}")
print(f"Upper: {text.upper()}")
# Output:
# Original: hello world
# Upper: HELLO WORLD
```

```python
# Example 2: lower() - converts all characters to lowercase
text = "HELLO WORLD"
print(f"Original: {text}")
print(f"Lower: {text.lower()}")
# Output:
# Original: HELLO WORLD
# Lower: hello world
```

```python
# Example 3: capitalize() - capitalizes first letter, rest lowercase
text = "hello WORLD"
print(f"Original: {text}")
print(f"Capitalize: {text.capitalize()}")
# Output:
# Original: hello WORLD
# Capitalize: Hello world
```

```python
# Example 4: Using all three methods together
text = "pYtHoN PrOgRaMmInG"
print(f"Original: {text}")
print(f"Upper: {text.upper()}")
print(f"Lower: {text.lower()}")
print(f"Capitalize: {text.capitalize()}")
# Output:
# Original: pYtHoN PrOgRaMmInG
# Upper: PYTHON PROGRAMMING
# Lower: python programming
# Capitalize: Python programming
```

```python
# Example 1: Split by default (whitespace)
text = "Hello World Python"
print(f"Original: {text}")
print(f"Split: {text.split()}")
# Output:
# Original: Hello World Python
# Split: ['Hello', 'World', 'Python']
```

```python
# Example 2: Split by specific character
text = "apple,banana,orange,grape"
print(f"Split by comma: {text.split(',')}")
# Output: Split by comma: ['apple', 'banana', 'orange', 'grape']
```

```python
# Example 3: Split by space explicitly
text = "one two three four"
print(f"Split by space: {text.split(' ')}")
# Output: Split by space: ['one', 'two', 'three', 'four']
```

```python
# Example 1: title() - converts first letter of each word to uppercase
text = "hello world"
print(f"Original: {text}")
print(f"Title: {text.title()}")
# Output:
# Original: hello world
# Title: Hello World

# Example 2: title() with mixed case
text = "pYtHoN pRoGrAmMiNg"
print(f"Original: {text}")
print(f"Title: {text.title()}")
# Output:
# Original: pYtHoN pRoGrAmMiNg
# Title: Python Programming
```

```python
# Example 1: Replace a word in a sentence
sentence = "I love Python programming"
print(f"Original: {sentence}")
print(f"Replaced: {sentence.replace('Python', 'Java')}")
# Output:
# Original: I love Python programming
# Replaced: I love Java programming

# Example 2: Replace multiple occurrences
sentence = "Python is great. Python is powerful. Python is fun."
print(f"Original: {sentence}")
print(f"Replaced: {sentence.replace('Python', 'JavaScript')}")
# Output:
# Original: Python is great. Python is powerful. Python is fun.
# Replaced: JavaScript is great. JavaScript is powerful. JavaScript is fun.
```

```python
# Example 1: Addition (+)
a = 10
b = 5
print(f"{a} + {b} = {a + b}")
# Output: 10 + 5 = 15

# Example 2: Subtraction (-)
a = 10
b = 5
print(f"{a} - {b} = {a - b}")
# Output: 10 - 5 = 5

# Example 3: Multiplication (*)
a = 10
b = 5
print(f"{a} * {b} = {a * b}")
# Output: 10 * 5 = 50

# Example 4: Division (/)
a = 10
b = 5
print(f"{a} / {b} = {a / b}")
# Output: 10 / 5 = 2.0

# Example 5: Floor Division (//)
a = 10
b = 3
print(f"{a} // {b} = {a // b}")  # Rounds down
# Output: 10 // 3 = 3

# Example 6: Modulus (%)
a = 10
b = 3
print(f"{a} % {b} = {a % b}")    # Remainder
# Output: 10 % 3 = 1

# Example 7: Exponentiation (**)
a = 10
b = 3
print(f"{a} ** {b} = {a ** b}")  # 10^3
# Output: 10 ** 3 = 1000
```

```python
# Example 1: Simple Assignment (=)
x = 10
print(f"x = {x}")
# Output: x = 10

# Example 2: Assign multiple variables
a = b = c = 5
print(f"a = {a}, b = {b}, c = {c}")
# Output: a = 5, b = 5, c = 5

# Example 3: Assign different values in one line
x, y, z = 10, 20, 30
print(f"x = {x}, y = {y}, z = {z}")
# Output: x = 10, y = 20, z = 30

# Example 4: Swapping variables
x = 10
y = 20
print(f"Before swap: x = {x}, y = {y}")
x, y = y, x  # Swap
print(f"After swap: x = {x}, y = {y}")
# Output:
# Before swap: x = 10, y = 20
# After swap: x = 20, y = 10
```

```python
# Example 5: Addition Assignment (+=)
x = 10
print(f"Initial: x = {x}")
x += 5  # x = x + 5
print(f"x += 5: x = {x}")
# Output:
# Initial: x = 10
# x += 5: x = 15

# Example 6: Subtraction Assignment (-=)
x = 10
print(f"Initial: x = {x}")
x -= 3  # x = x - 3
print(f"x -= 3: x = {x}")
# Output:
# Initial: x = 10
# x -= 3: x = 7

# Example 7: Multiplication Assignment (*=)
x = 10
print(f"Initial: x = {x}")
x *= 2  # x = x * 2
print(f"x *= 2: x = {x}")
# Output:
# Initial: x = 10
# x *= 2: x = 20

# Example 8: Division Assignment (/=)
x = 10
print(f"Initial: x = {x}")
x /= 4  # x = x / 4
print(f"x /= 4: x = {x}")
# Output:
# Initial: x = 10
# x /= 4: x = 2.5

# Example 9: Floor Division Assignment (//=)
x = 10
print(f"Initial: x = {x}")
x //= 4  # x = x // 4
print(f"x //= 4: x = {x}")
# Output:
# Initial: x = 10
# x //= 4: x = 2

# Example 10: Modulus Assignment (%=)
x = 10
print(f"Initial: x = {x}")
x %= 3  # x = x % 3
print(f"x %= 3: x = {x}")
# Output:
# Initial: x = 10
# x %= 3: x = 1

# Example 11: Exponentiation Assignment (**=)
x = 10
print(f"Initial: x = {x}")
x **= 2  # x = x ** 2
print(f"x **= 2: x = {x}")
# Output:
# Initial: x = 10
# x **= 2: x = 100
```
