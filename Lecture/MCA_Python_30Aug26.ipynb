```python
# Example 1: Counting characters in a string
text = "Hello World"
print(f"Length of '{text}': {len(text)}")
# Output: Length of 'Hello World': 11
```

```python
# Example 2: Counting specific characters
text = "banana"
print(f"Count of 'a' in '{text}': {text.count('a')}")
# Output: Count of 'a' in 'banana': 3
```

```python
sentence = "Python is awesome and Python is fun"
words = sentence.split()
print(f"Number of words: {len(words)}")
# Output: Number of words: 7
```

```python
# Example 4: Counting specific words
print(f"Count of 'Python': {sentence.count('Python')}")
# Output: Count of 'Python': 2
```

```python
# Example 1: upper() - converts all characters to uppercase
text = "hello world"
print(f"Original: {text}")
print(f"Upper: {text.upper()}")
# Output:
# Original: hello world
# Upper: HELLO WORLD
```

```python
# Example 2: lower() - converts all characters to lowercase
text = "HELLO WORLD"
print(f"Original: {text}")
print(f"Lower: {text.lower()}")
# Output:
# Original: HELLO WORLD
# Lower: hello world
```

```python
# Example 3: capitalize() - capitalizes first letter, rest lowercase
text = "hello WORLD"
print(f"Original: {text}")
print(f"Capitalize: {text.capitalize()}")
# Output:
# Original: hello WORLD
# Capitalize: Hello world
```

```python
# Example 4: Using all three methods together
text = "pYtHoN PrOgRaMmInG"
print(f"Original: {text}")
print(f"Upper: {text.upper()}")
print(f"Lower: {text.lower()}")
print(f"Capitalize: {text.capitalize()}")
# Output:
# Original: pYtHoN PrOgRaMmInG
# Upper: PYTHON PROGRAMMING
# Lower: python programming
# Capitalize: Python programming
```

```python
# Example 1: Split by default (whitespace)
text = "Hello World Python"
print(f"Original: {text}")
print(f"Split: {text.split()}")
# Output:
# Original: Hello World Python
# Split: ['Hello', 'World', 'Python']
```

```python
# Example 2: Split by specific character
text = "apple,banana,orange,grape"
print(f"Split by comma: {text.split(',')}")
# Output: Split by comma: ['apple', 'banana', 'orange', 'grape']
```

```python
# Example 3: Split by space explicitly
text = "one two three four"
print(f"Split by space: {text.split(' ')}")
# Output: Split by space: ['one', 'two', 'three', 'four']
```

```python
# Example 1: title() - converts first letter of each word to uppercase
text = "hello world"
print(f"Original: {text}")
print(f"Title: {text.title()}")
# Output:
# Original: hello world
# Title: Hello World

# Example 2: title() with mixed case
text = "pYtHoN pRoGrAmMiNg"
print(f"Original: {text}")
print(f"Title: {text.title()}")
# Output:
# Original: pYtHoN pRoGrAmMiNg
# Title: Python Programming
```

```python
# Example 1: Replace a word in a sentence
sentence = "I love Python programming"
print(f"Original: {sentence}")
print(f"Replaced: {sentence.replace('Python', 'Java')}")
# Output:
# Original: I love Python programming
# Replaced: I love Java programming

# Example 2: Replace multiple occurrences
sentence = "Python is great. Python is powerful. Python is fun."
print(f"Original: {sentence}")
print(f"Replaced: {sentence.replace('Python', 'JavaScript')}")
# Output:
# Original: Python is great. Python is powerful. Python is fun.
# Replaced: JavaScript is great. JavaScript is powerful. JavaScript is fun.
```
