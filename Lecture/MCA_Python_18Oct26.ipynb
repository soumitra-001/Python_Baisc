**Lambda**

```python
add = lambda a,b : a+b
```

```python
print(add(5,2))
```

```python
nums = [1,2,3,4]
squares = list(map(lambda x: x*x,nums))

print(squares)
```

```python
nums = [1,2,3,4,5,6]
evens = list(filter(lambda x : x%2==0,nums))
print(evens)
```

```python
from functools import reduce

nums = [1,2,3,4]

prod = reduce(lambda a,b : a*b , nums)

print(prod)
```

```python
vals = [-3,1,-10,2]

print(sorted(vals,key = lambda x : abs(x)))

#sorting based on absolute value
```

```python
person = [{"name" : "Riya" , "age" : 32},{"name" : "Ram" , "age" : 26}]


```

```python
print(sorted(person,key = lambda x : x["age"]))
```

```python
nums = [1,2,3,4]

print(min(nums))
print(max(nums))
```

```python
words = ["rambo","ram","ratsay"]

print(min(words, key = lambda x : len(x))) # shortest word length
print(max(words, key = lambda x : len(x))) # longest word length
```
