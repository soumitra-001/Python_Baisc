**Lambda**

```python
add = lambda a,b : a+b
```

```python
print(add(5,2))
```

```python
nums = [1,2,3,4]
squares = list(map(lambda x: x*x,nums))

print(squares)
```

```python
nums = [1,2,3,4,5,6]
evens = list(filter(lambda x : x%2==0,nums))
print(evens)
```

```python
from functools import reduce

nums = [1,2,3,4]

prod = reduce(lambda a,b : a*b , nums)

print(prod)
```

```python
vals = [-3,1,-10,2]

print(sorted(vals,key = lambda x : abs(x)))

#sorting based on absolute value
```

```python
person = [{"name" : "Riya" , "age" : 32},{"name" : "Ram" , "age" : 26}]


```

```python
print(sorted(person,key = lambda x : x["age"]))
```

```python
nums = [1,2,3,4]

print(min(nums))
print(max(nums))
```

```python
words = ["rambo","ram","ratsay"]

print(min(words, key = lambda x : len(x))) # shortest word length
print(max(words, key = lambda x : len(x))) # longest word length
```

```python
# conditional expression inside lambda

sign = lambda x : "zero" if x == 0 else ("posative" if x>0 else "negative")

print(sign(-1),sign(0),sign(4))
```

```python
# 2 lambda fuction

make_pow = lambda n : (lambda x : x**n)
square = make_pow(2)
cube = make_pow(3)

print(square(5))
print(cube(3))
```

```python
# pair wise sorting

data = [(1, 5), (3, 2), (2, 8), (1, 3), (3, 1)]

result = sorted(data, key=lambda x: x[0])

print(result)
```
